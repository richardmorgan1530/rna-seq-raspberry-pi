**Trim bad data from the fastq.gz files using trimmomatic:**

```python
%%bash

INPUT_DIR="/var/www/html/usb/jupyter/my_projects/data/SRP093363/fastq"

# Change the dir name to:
#INPUT_DIR="/var/www/html/usb/jupyter/my_projects/data/SRP093363/fastq_gz"

OUTPUT_DIR="/var/www/html/usb/jupyter/my_projects/data/SRP093363/fastq-trimmed"

mkdir -p "$OUTPUT_DIR"

echo "Using Trimmomatic:"
which trimmomatic

for FILE in "$INPUT_DIR"/*.fastq.gz
do
    BASENAME=$(basename "$FILE")

    echo "Processing $BASENAME"

    trimmomatic SE -threads 2 \
        "$FILE" \
        "$OUTPUT_DIR/$BASENAME" \
        SLIDINGWINDOW:4:20 MINLEN:75
done

echo "Trimming complete!"
```
