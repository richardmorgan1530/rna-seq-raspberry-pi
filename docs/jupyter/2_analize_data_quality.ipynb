**Analyze fastq.gz files using fastqc command:**

This will produce a .html report file containing analysis results and a .zip file which contains the analysis images and results

```python
%%bash

# Test code only, use the next code for actual runs:
# fastqc /var/www/html/usb/jupyter/my_projects/data/SRP093363/fastq/SRR5017135.fastq.gz
```

```python
%%bash

FASTQ_DIR="/var/www/html/usb/jupyter/my_projects/data/SRP093363/fastq"

# Change the dir name to:
# FASTQ_DIR="/var/www/html/usb/jupyter/my_projects/data/SRP093363/fastq_gz"

FASTQC_DIR="/var/www/html/usb/jupyter/my_projects/data/SRP093363/fastqc-untrimmed-results"

# Create FastQC output folder
mkdir -p "$FASTQC_DIR"

find "$FASTQ_DIR" -name "*.fastq.gz" | while read FILE; do

    BASENAME=$(basename "$FILE")

    echo "Running FastQC on $BASENAME"

    fastqc \
        "$FILE" \
        --outdir "$FASTQC_DIR"

done
```
